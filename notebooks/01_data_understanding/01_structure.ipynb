## House Prices (Estructura de los datos)

```python
from house_prices_mlops.pipeline import load_stage
from house_prices_mlops.data_tracker import DataTracker


df_hp = load_stage("train")
tracker = DataTracker.load(df_hp)
```

```python
#Shape
display(df_hp.shape)
```

### Variables Types

```python
display(tracker.profile)
```

```python
# DU — Structure: revisión de naturaleza y representación técnica

structure = tracker.profile[
    ["var", "variable_type", "data_type", "unique_count"]
].copy()

variable_groups = [
    "continuous",
    "discrete",
    "ordinal",
    "nominal",
    "temporal",
    "identifier"
]

for kind in variable_groups:
    group = structure[structure["variable_type"] == kind]
    print(f"{kind.upper()} — {len(group)} variables")
    display(group.reset_index(drop=True))
```

### Cardinality (Naturaleza de la variable)

```python
# Cardinalidad de variables categóricas

cardinality_review = tracker.profile[
    tracker.profile['variable_type'].isin([
        'nominal',
        'ordinal'
    ])
][
    ['var', 'data_type', 'variable_type', 'unique_count']
].sort_values(
    'unique_count',
    ascending=False
)

display(cardinality_review)
```

```python
tracker.add_finding(key='ms_subclass_representation', 
    var="MSSubClass",
    stage="DU_Structure",
    finding=(
        "Variable nominal representada mediante códigos enteros. "
        "Su tratamiento en DP debe preservar el carácter categórico."
    )
)
tracker.save()
```
